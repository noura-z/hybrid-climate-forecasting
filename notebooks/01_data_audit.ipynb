```python
import pandas as pd

file_path = "../data/raw/All data.xlsx"

df = pd.read_excel(file_path)

print(df.shape)
df.head()
```

```python
df.info()
```

```python
column_mapping = {
    "Local time in Rabat (airport)": "timestamp",
    "T": "temperature_c",
    "Po": "station_pressure_hpa",
    "P": "sea_level_pressure_hpa",
    "Pa": "pressure_change_hpa",
    "U": "humidity_pct",
    "DD": "wind_direction",
    "Ff": "wind_speed_ms",
    "ff10": "wind_gust_10min_ms",
    "ff3": "wind_gust_3sec_ms",
    "N": "cloud_cover",
    "WW": "present_weather_code",
    "W1": "past_weather_code_1",
    "W2": "past_weather_code_2",
    "Tn": "min_temperature_c",
    "Tx": "max_temperature_c",
    "Cl": "low_cloud_type",
    "Nh": "cloud_amount",
    "H": "cloud_base_height",
    "Cm": "middle_cloud_type",
    "Ch": "high_cloud_type",
    "VV": "visibility",
    "Td": "dew_point_c",
    "RRR": "precipitation_mm",
    "tR": "precipitation_period_h",
    "E": "ground_condition",
    "Tg": "ground_temperature_c",
    "E'": "ground_state",
    "sss": "snow_depth"
}
df = df.rename(columns=column_mapping)
```

```python
columns_to_drop = [
    "ground_condition",
    "snow_depth",
    "ground_state",
    "ground_temperature_c",
    "wind_gust_10min_ms",
    "wind_gust_3sec_ms",
    "past_weather_code_1",
    "past_weather_code_2"
]

df = df.drop(columns=columns_to_drop)

print("Remaining columns:", len(df.columns))
```

```python
df["precipitation_mm"].value_counts(dropna=False)
```

```python
df["precipitation_mm"] = (
    df["precipitation_mm"]
      .replace("No precipitation", 0)
      .replace("Trace of precipitation", 0.01)
)

df["precipitation_mm"] = pd.to_numeric(
    df["precipitation_mm"],
    errors="coerce"
)

df["precipitation_mm"] = df["precipitation_mm"].fillna(0)

df["precipitation_mm"].describe()
```

```python
df["precipitation_period_h"] = (
    df["precipitation_period_h"]
      .fillna(0)
)
```

```python
continuous_cols = [
    "temperature_c",
    "humidity_pct",
    "station_pressure_hpa",
    "sea_level_pressure_hpa",
    "pressure_change_hpa",
    "wind_speed_ms",
    "dew_point_c",
    "visibility"
]
for col in continuous_cols:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

for col in continuous_cols:
    df[col] = df[col].interpolate()
```

```python
df[[
    "temperature_c",
    "min_temperature_c",
    "max_temperature_c"
]].head(30)
```

```python
df = df.drop(
    columns=[
        "min_temperature_c",
        "max_temperature_c"
    ]
)

print(df.shape)
```

```python
df = df.drop(
    columns=[
        "high_cloud_type",
        "middle_cloud_type",
        "low_cloud_type"
    ]
)
```

```python
df = df.drop(
    columns=[
        "cloud_amount",
        "cloud_base_height"
    ]
)
```

```python
sorted(df["cloud_cover"].dropna().unique())
df["cloud_cover"].value_counts(dropna=False)
```

```python
# Normalize cloud cover strings

df["cloud_cover"] = (
    df["cloud_cover"]
      .astype(str)
      .str.strip()
      .str.replace("–", "-", regex=False)
      .str.replace("  ", " ", regex=False)
)

cloud_mapping = {
    "no clouds": 0,
    "10% or less, but not 0": 1,
    "20-30%": 2,
    "40%": 3,
    "50%": 4,
    "60%": 5,
    "70 - 80%": 6,
    "90 or more, but not 100%": 7,
    "100%": 8
}

df["cloud_cover_encoded"] = (
    df["cloud_cover"]
    .map(cloud_mapping)
)
```

```python
df["cloud_cover_encoded"] = (
    df["cloud_cover_encoded"]
    .fillna(df["cloud_cover_encoded"].median())
)
```

```python
df = df.drop(columns=["cloud_cover"])
```

```python
df["wind_direction"] = (
    df["wind_direction"]
      .replace("ffill", pd.NA)
      .ffill()
)
```

```python
sorted(df["wind_direction"].dropna().unique())
```

```python
wind_mapping = {
    "Calm, no wind": 0,
    "Wind blowing from the north": 0,
    "Wind blowing from the north-northeast": 22.5,
    "Wind blowing from the north-east": 45,
    "Wind blowing from the east-northeast": 67.5,
    "Wind blowing from the east": 90,
    "Wind blowing from the east-southeast": 112.5,
    "Wind blowing from the south-east": 135,
    "Wind blowing from the south-southeast": 157.5,
    "Wind blowing from the south": 180,
    "Wind blowing from the south-southwest": 202.5,
    "Wind blowing from the south-west": 225,
    "Wind blowing from the west-southwest": 247.5,
    "Wind blowing from the west": 270,
    "Wind blowing from the west-northwest": 292.5,
    "Wind blowing from the north-west": 315,
    "Wind blowing from the north-northwest": 337.5
}
df["wind_direction_deg"] = (
    df["wind_direction"]
    .map(wind_mapping)
)
```

```python
import numpy as np

radians = np.deg2rad(df["wind_direction_deg"])

df["wind_dir_sin"] = np.sin(radians)
df["wind_dir_cos"] = np.cos(radians)
```

```python
df=df.drop(columns=["wind_direction","wind_direction_deg"])
```
